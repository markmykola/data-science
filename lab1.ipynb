```python
import numpy as np

# --- Вхідні дані ---
n = 4  # Введи сюди свій номер за списком!
size = n + 2

print(f"Варіант (n): {n}")
print(f"Розмірність матриць: {size}x{size}")

# 1. Створення двох двовимірних масивів з випадковими числами від 0 до 9 [cite: 24, 280]
# Використовуємо np.random.randint для цілих чисел
matrix_1 = np.random.randint(0, 10, (size, size))
matrix_2 = np.random.randint(0, 10, (size, size))

print("\nМатриця 1:\n", matrix_1)
print("\nМатриця 2:\n", matrix_2)

# 2. Арифметичні операції [cite: 49-57, 281]
add_res = np.add(matrix_1, matrix_2)      # Додавання
sub_res = np.subtract(matrix_1, matrix_2) # Віднімання
# Додаємо маленьке число, щоб уникнути ділення на нуль
div_res = np.divide(matrix_1, matrix_2 + 1e-10) 

# Скалярне множення (матричний добуток) [cite: 58, 59]
dot_res = np.dot(matrix_1, matrix_2)

# Векторне множення (поелементне) [cite: 54, 55]
mult_res = np.multiply(matrix_1, matrix_2)

print("\n--- Результати операцій ---")
print("Сума (перший рядок):\n", add_res[0])
print("Скалярний добуток (dot):\n", dot_res[0])
print("Поелементне множення:\n", mult_res[0])

# 3. Лінійна алгебра для matrix_1 [cite: 68-75, 282]
from scipy import linalg

det = linalg.det(matrix_1) # Визначник
try:
    inv = linalg.inv(matrix_1) # Обернена матриця
except linalg.LinAlgError:
    inv = "Матриця вироджена (визначник = 0)"

eigenvalues, eigenvectors = linalg.eig(matrix_1) # Власні числа та вектори

print("\n--- Лінійна алгебра (для Матриці 1) ---")
print(f"Визначник: {det:.2f}")
print("Власні числа:\n", eigenvalues)
```

```python
from scipy import optimize, integrate
from scipy.misc import derivative
import numpy as np

# Визначення функції з Варіанту 4 (Таблиця 1.1)
def f(x):
    # ln^2(x) / x -> (log(x)**2) / x
    return (np.log(x)**2) / x

point = 2.0
interval = [1.0, 4.0]

# 1. Похідна в точці x = 2.0 [cite: 79-84]
# dx - крок для обчислення похідної
der = derivative(f, point, dx=1e-6)
print(f"Похідна в точці {point}: {der:.5f}")

# 2. Первісна (визначений інтеграл) на проміжку [1.0, 4.0] [cite: 85-88]
integral_val, error = integrate.quad(f, interval[0], interval[1])
print(f"Інтеграл на проміжку {interval}: {integral_val:.5f}")

# 3. Оптимізація (мінімуми) [cite: 94-103]
# Шукаємо локальний мінімум, починаючи з точки 0.5 (зліва) та 3.0 (справа)
res_min_global = optimize.minimize(f, x0=0.5, bounds=[(0.1, 5)]) 
res_min_local = optimize.minimize(f, x0=3.0, bounds=[(0.1, 5)])

# Для наочності знайдемо максимум (мінімізуємо -f(x))
def neg_f(x): return -f(x)
res_max = optimize.minimize(neg_f, x0=2.0, bounds=[(0.1, 5)])

print(f"Глобальний мінімум (x): {res_min_global.x[0]:.2f}, f(x): {res_min_global.fun:.2f}")
print(f"Максимум функції (x): {res_max.x[0]:.2f} (Приблизно e -> 2.718)")
```

```python
import pandas as pd
import numpy as np

# 1. Завантаження [cite: 145]
df = pd.read_csv('titanic.csv')

print("--- ЗАВДАННЯ 6: Базові операції ---")

# 2. Огляд даних (Head/Tail/Shape) [cite: 148-152]
print(f"Розмірність датасету: {df.shape}")
print("\nПерші 3 рядки:\n", df.head(3))
print("\nОстанні 3 рядки:\n", df.tail(3))

# 3. Типи даних та колонки [cite: 154-156]
print("\nТипи даних:\n", df.dtypes)
```

```python

# 4. Вибір конкретних стовпців [cite: 159]
print("\nТільки імена та клас каюти:\n", df[['Name', 'Pclass']].head())

# 5. Індексація loc та iloc [cite: 160-171]
print("\nДані 10-го пасажира (iloc):\n", df.iloc[10])
print("\nЗріз даних (loc): рядки 5-8, стовпці Name та Age:\n", df.loc[5:8, ['Name', 'Age']])
```

```python

# 6. Фільтрація (Маски) [cite: 172-175]
# Шукаємо жінок (Sex== 'female') старших за 50 років
filtered_df = df[(df['Sex'] == 'female') & (df['Age'] > 50)]
print(f"\nКількість жінок старше 50 років: {len(filtered_df)}")

# 7. Статистика (Describe, Value Counts) [cite: 176-183]
print("\nСтатистика по числових полях:\n", df.describe())
print("\nКількість пасажирів у кожному класі:\n", df['Pclass'].value_counts())
```

```python

print("\n--- ЗАВДАННЯ 7: Групування та Перетворення ---")

# 8. Групування (Groupby) [cite: 198-215]
# Групуємо за класом каюти (Pclass) та рахуємо середню вартість квитка (Fare) і середній вік (Age)
grouped = df.groupby(['Pclass'])[['Fare', 'Age']].mean()
print("\nСередня ціна та вік по класах:\n", grouped)

# Агрегація з кількома функціями (agg) [cite: 220-224]
# Для віку шукаємо мінімум і максимум у кожному класі
agg_res = df.groupby(['Pclass']).agg({'Age': ['min', 'max'], 'Fare': 'mean'})
print("\nМін. і макс. вік по класах:\n", agg_res)

# 9. Перетворення ознак (Map/Apply) [cite: 265-275]
# map: Перетворимо текстову стать у числову (male=0, female=1)
sex_map = {'male': 0, 'female': 1}
df['Sex_Code'] = df['Sex'].map(sex_map)

# apply: Конвертуємо ціну квитка з фунтів у долари (умовно * 1.3)
df['Fare_USD'] = df['Fare'].apply(lambda x: x * 1.3)

print("\nРезультат перетворення (нові стовпці):\n", df[['Name', 'Sex', 'Sex_Code', 'Fare', 'Fare_USD']].head())
```

```python
import pandas as pd

# 1. Завантаження трьох таблиць [cite: 235]
# calendar.csv - доступність житла по датах
# listings.csv - опис квартир
# reviews.csv - відгуки користувачів
try:
    calendar = pd.read_csv('calendar.csv')
    listings = pd.read_csv('listings.csv')
    reviews = pd.read_csv('reviews.csv')
    print("Файли успішно завантажено!")
except FileNotFoundError:
    print("Помилка: Переконайтеся, що файли (calendar.csv, listings.csv, reviews.csv) знаходяться в папці.")

# 2. Попередній огляд
print("Listings shape:", listings.shape)
print("Reviews shape:", reviews.shape)

# --- MERGE (Злиття) --- [cite: 242-249]
# З'єднаємо описи квартир (listings) з відгуками про них (reviews).
# У listings ключ 'id', у reviews ключ 'listing_id'.
# Використовуємо 'inner', щоб отримати тільки ті квартири, які мають відгуки.

merged_data = pd.merge(
    listings[['id', 'name', 'price', 'neighbourhood']], # беремо тільки потрібні стовпці
    reviews[['listing_id', 'reviewer_name', 'comments']], 
    left_on='id', 
    right_on='listing_id', 
    how='inner'
)

print("\n--- Результат MERGE (Квартири + Відгуки) ---")
print(merged_data.head())


# --- JOIN (Приєднання за індексом) --- [cite: 252-256]
# Для join нам потрібно встановити індекси.
# Давайте приєднаємо календар (ціни та дати) до лістингів.

# Встановлюємо 'id' як індекс для listings
listings_indexed = listings.set_index('id')

# Встановлюємо 'listing_id' як індекс для calendar
calendar_indexed = calendar.set_index('listing_id')

# Об'єднуємо. Оскільки рядків дуже багато, візьмемо тільки перші 100 для прикладу
# lsuffix/rsuffix потрібні, якщо є однакові назви колонок
joined_data = listings_indexed.head(100).join(
    calendar_indexed.head(100), 
    lsuffix='_info', 
    rsuffix='_schedule'
)

print("\n--- Результат JOIN (Квартири + Календар) ---")
# Виводимо назву квартири (з лівої таблиці) і дату/доступність (з правої)
print(joined_data[['name', 'date', 'available']].head())
```

```python
import pandas as pd

# Завантаження таблиць
try:
    drivers = pd.read_csv('drivers.csv')
    races = pd.read_csv('races.csv')
    results = pd.read_csv('results.csv')
    print("Файли Формули-1 завантажено.")
except FileNotFoundError:
    print("Помилка: переконайтеся, що файли drivers.csv, races.csv, results.csv є в папці.")

# Огляд ключів
# drivers має 'driverId'
# races має 'raceId'
# results має і 'driverId', і 'raceId' — це зв'язуюча ланка

# [cite_start]--- MERGE (Об'єднання таблиць) [cite: 227-249] ---
# Завдання: Дізнатися імена водіїв, які приїхали першими (positionOrder = 1)
# З'єднуємо results (результати) з drivers (імена) по 'driverId'

# Спочатку відфільтруємо тільки переможців, щоб таблиця була меншою
winners_only = results[results['positionOrder'] == 1]

# Merge
winners_names = pd.merge(
    winners_only[['resultId', 'raceId', 'driverId']], # Ліва таблиця
    drivers[['driverId', 'forename', 'surname', 'nationality']], # Права таблиця
    on='driverId', # Ключ
    how='inner'
)

print("\n--- Результат MERGE (Переможці гонок) ---")
print(winners_names.head())


# [cite_start]--- JOIN (Приєднання за індексом) [cite: 251-258] ---
# Завдання: Додати до таблиці переможців назву траси та рік.
# Використовуємо 'raceId' як індекс для з'єднання.

# Встановлюємо індекси
winners_indexed = winners_names.set_index('raceId')
races_indexed = races.set_index('raceId')

# Join: до результатів додаємо інформацію про гонку
full_info = winners_indexed.join(
    races_indexed[['year', 'name']], # Беремо тільки рік і назву траси
    rsuffix='_race' # Суфікс на випадок однакових колонок (хоча тут їх немає)
)

print("\n--- Результат JOIN (Переможець + Назва траси) ---")
# Виводимо рік, трасу і ім'я переможця
print(full_info[['year', 'name', 'forename', 'surname']].sort_values('year').tail())
```

```python
import pandas as pd

# Налаштування для гарного відображення таблиць
pd.set_option('display.max_columns', None)
pd.set_option('display.width', 1000)

print("\n" + "="*50)
print("   ЗАВДАННЯ 8: MOVIELENS (ФІЛЬМИ + РЕЙТИНГИ)")
print("="*50)

# 1. Завантаження даних
try:
    movies = pd.read_csv('movies.csv')
    ratings = pd.read_csv('ratings.csv')
    links = pd.read_csv('links.csv')
    print(">>> Файли movies, ratings та links успішно завантажено.")
except FileNotFoundError:
    print("!!! Помилка: Не знайдено файли (movies.csv, ratings.csv, links.csv).")

# Огляд розмірності
print(f"Фільмів: {movies.shape[0]}, Рейтингів: {ratings.shape[0]}")

# --- 2. MERGE (Об'єднання за спільним стовпцем) [cite: 227-249] ---
# У таблиці 'ratings' є тільки ID фільму, але немає назви.
# У таблиці 'movies' є ID і назва.
# Давайте об'єднаємо їх, щоб бачити, якому фільму поставили 5 зірок.

# Беремо тільки високі оцінки (5.0), щоб таблиця була меншою
top_ratings = ratings[ratings['rating'] == 5.0].head(100) # Беремо перші 100 п'ятірок

# Об'єднуємо (merge) таблицю оцінок з таблицею фільмів по 'movieId'
merged_movies = pd.merge(
    top_ratings[['userId', 'movieId', 'rating']], # Ліва таблиця (оцінки)
    movies[['movieId', 'title', 'genres']],       # Права таблиця (назви)
    on='movieId',                                 # Ключ
    how='inner'
)

print("\n>>> Результат MERGE (Користувачі та їхні улюблені фільми):")
print(merged_movies[['userId', 'rating', 'title']].head())


# --- 3. JOIN (Приєднання за індексом)  ---
# У нас є таблиця 'links', яка містить посилання на IMDB.
# Давайте приєднаємо ці посилання до нашої таблиці фільмів 'movies'.
# Для join нам потрібні однакові індекси.

# Встановлюємо 'movieId' як індекс для обох таблиць
movies_indexed = movies.set_index('movieId')
links_indexed = links.set_index('movieId')

# Приєднуємо links до movies
# Використовуємо lsuffix/rsuffix, хоча тут колонки унікальні, це хороша практика
movies_with_links = movies_indexed.join(
    links_indexed,
    lsuffix='_mov',
    rsuffix='_lnk'
)

print("\n>>> Результат JOIN (Фільми + ID з IMDB):")
# Виводимо назву і код IMDB для перших 5 фільмів
print(movies_with_links[['title', 'imdbId', 'tmdbId']].head())

# Додатково: збережемо результат для звіту
movies_with_links.head(20).to_csv('movies_final_report.csv')
print("\n>>> Звіт збережено у 'movies_final_report.csv'")
```
